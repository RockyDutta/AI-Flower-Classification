# AI Flower Classification — Exploratory Analysis

This notebook walks through the same pipeline implemented in `src/`, used here for interactive exploration and sanity-checking before productionizing into `train.py`.

```python
import sys, os
sys.path.append(os.path.join(os.getcwd(), '..', 'src'))

from data_loader import load_iris_dataframe, dataset_summary, get_feature_target_split, FEATURE_NAMES
from preprocessing import drop_duplicates, split_data, scale_features

df = load_iris_dataframe()
df.head()
```

```python
dataset_summary(df)
```

```python
df_clean = drop_duplicates(df)
x, y = get_feature_target_split(df_clean)
x_train, x_test, y_train, y_test = split_data(x, y, test_size=0.2, random_state=42)
x_train_scaled, x_test_scaled, scaler = scale_features(x_train, x_test)
x_train_scaled.describe()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report

model = KNeighborsClassifier(n_neighbors=9)
model.fit(x_train_scaled, y_train)
preds = model.predict(x_test_scaled)

print('Accuracy:', accuracy_score(y_test, preds))
print(classification_report(y_test, preds))
```
